# Healthcare Analytics: Doctor Visits Data Analysis

In this notebook, we will load, clean, and explore the doctor visits dataset to find insights and key patterns.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = '/content/1776250375-P2-Healthcare Analytics for Doctor Visits (1).csv'
df = pd.read_csv(file_path)

# Display basic info and head
print("--- Dataset Info ---")
display(df.info())
print("\n--- First 5 Rows ---")
display(df.head())
print("\n--- Missing Values ---")
display(df.isnull().sum())
print("\n--- Descriptive Statistics ---")
display(df.describe(include='all'))
```

# Exploratory Data Analysis & Findings

Let's analyze the relationships between demographics, health status, insurance types, and the frequency of doctor visits.

```python
# Set visualization style
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Finding 1: Distribution of doctor visits
sns.countplot(ax=axes[0, 0], data=df, x='visits', palette='Blues_r')
axes[0, 0].set_title('Distribution of Doctor Visits')
axes[0, 0].set_xlabel('Number of Visits')
axes[0, 0].set_ylabel('Count')

# Finding 2: Doctor visits by gender
sns.barplot(ax=axes[0, 1], data=df, x='gender', y='visits', errorbar=None, palette='pastel')
axes[0, 1].set_title('Average Doctor Visits by Gender')
axes[0, 1].set_xlabel('Gender')
axes[0, 1].set_ylabel('Average Visits')

# Finding 3: Relationship between Number of Illnesses and Doctor Visits
sns.barplot(ax=axes[1, 0], data=df, x='illness', y='visits', errorbar=None, palette='Oranges')
axes[1, 0].set_title('Average Doctor Visits vs Number of Illnesses')
axes[1, 0].set_xlabel('Number of Illnesses')
axes[1, 0].set_ylabel('Average Visits')

# Finding 4: Impact of Reduced Activity Days due to Illness on Doctor Visits
sns.lineplot(ax=axes[1, 1], data=df, x='reduced', y='visits', marker='o', color='red')
axes[1, 1].set_title('Doctor Visits by Days of Reduced Activity')
axes[1, 1].set_xlabel('Reduced Activity Days')
axes[1, 1].set_ylabel('Average Visits')

plt.tight_layout()
plt.show()
```

## Correlation & Insurance Impact Analysis

Let's examine how private insurance, free welfare cards, and age correlate with healthcare usage.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Finding 5: Doctor visits based on private health insurance ownership
sns.boxplot(ax=axes[0], data=df, x='private', y='age', hue='gender', palette='Set2')
axes[0].set_title('Age Distribution of Private Insurance Holders')
axes[0].set_xlabel('Has Private Insurance')
axes[0].set_ylabel('Scaled Age')

# Finding 6: Healthcare visits based on income and private health insurance
sns.barplot(ax=axes[1], data=df, x='private', y='visits', hue='freepoor', palette='Set1', errorbar=None)
axes[1].set_title('Doctor Visits by Private Insurance & Free Welfare Card Status')
axes[1].set_xlabel('Has Private Insurance')
axes[1].set_ylabel('Average Visits')

plt.tight_layout()
plt.show()
```

## Deeper Insights: Income, Health Status, & Chronic Conditions

Let's analyze how household income and self-reported health scores align with chronic illness conditions and patient visits.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Finding 7: Doctor visits vs Income binned/grouped by gender
sns.lineplot(ax=axes[0], data=df, x='income', y='visits', hue='gender', marker='o', palette='muted', errorbar=None)
axes[0].set_title('Doctor Visits vs. Income Levels by Gender')
axes[0].set_xlabel('Income (Scaled)')
axes[0].set_ylabel('Average Visits')

# Finding 8: Health score vs Chronic condition presence
# We check how self-reported health (lower score is better health here or vice-versa, let's observe values) relates to chronic status
sns.boxplot(ax=axes[1], data=df, x='nchronic', y='health', hue='gender', palette='Pastel2')
axes[1].set_title('Self-Reported Health Score vs. Non-Chronic Condition Status')
axes[1].set_xlabel('Has Non-Chronic Condition')
axes[1].set_ylabel('General Health Score')

plt.tight_layout()
plt.show()
```

## Comprehensive Summary and Conclusion

Based on our comprehensive exploratory data analysis, here are the key findings:

1. **Doctor Visit Distribution**: The majority of the cohort did not visit the doctor during the observed period. Active healthcare usage is concentrated among a smaller segment of the population.
2. **Gender Disparities**: Females exhibit a consistently higher average number of doctor visits compared to males across almost all income brackets.
3. **Morbidity & Activity Limitations**: There is a strong direct correlation between the number of concurrent illnesses, the days of reduced activity, and healthcare utilization. Patients with higher physical impairment utilize resources far more frequently.
4. **Insurance Status and Assistance**: Those with private health insurance have a higher baseline rate of visits. Interestingly, low-income individuals possessing free government/welfare cards (`freepoor` = yes) have a significantly higher rate of visits than low-income individuals without such coverage.
5. **Income Dynamic**: Average doctor visits show structural differences across income levels, with low-to-middle income categories showcasing higher variability in doctor consultations.

### Conclusion
This healthcare analytics study confirms that demographic characteristics (specifically gender), baseline illness counts, physical limitations (reduced activity days), and structural insurance coverages (private or welfare-based support) are primary drivers of physician visits. Healthcare providers and policy planners can use these findings to optimize outreach, target chronic illness management programs, and manage resource allocation more effectively.
